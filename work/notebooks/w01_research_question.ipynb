## 2. The question: decision, action, cost of a wrong call

*What decision does your work improve? Who acts on it? What does a wrong recommendation cost?*

Research question

Which pages should be reviewed first for a potential content refresh, based on observable search and content signals?

Decision

The decision is which pages should receive limited content-review attention first.

Action

For high-priority pages, a reviewer could inspect the page and decide whether to:

refresh or update the content;
expand thin content;
investigate CTR issues;
protect a page that is performing well;
monitor it instead of making a change.
Cost of a wrong recommendation

A false positive could cause a team to spend time reviewing or changing a page that did not need attention.

A false negative could be more costly because a genuinely declining or valuable page might be missed.

Therefore, I want the ranking to prioritize useful candidates while keeping the reasoning understandable to a human reviewer.
